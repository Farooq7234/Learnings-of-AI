# Create embeddings for documents

This lesson converts several text documents into vectors using OpenAI embeddings. Document vectors are typically stored in a vector index and compared with a query vector during semantic search.

## Before you begin
Install the project requirements and set `OPENAI_API_KEY` in `.env` or your environment. Calling the embedding API may incur usage charges.

## Steps
1. Import the embedding wrapper and load the API key from `.env`.
2. Configure the embedding model and request 32 output dimensions.
3. Put the example texts in a list. Each list item is treated as a separate document.
4. Call `embed_documents()` to get one vector per document, in the same order as the inputs.
5. Print the vectors. For a large collection, store them in a vector database instead of printing them.

## Context
The returned value is a list of vectors: vector 1 corresponds to the first text, vector 2 to the second, and so on. Each vector has 32 numbers because that dimension was requested. In a retrieval system, these vectors are stored with their source texts, then compared against a query embedding to find relevant documents.

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embedding = OpenAIEmbeddings(model='text-embedding-3-large', dimensions=32)

documents = [
    "Delhi is the capital of India",
    "Kolkata is the capital of West Bengal",
    "Paris is the capital of France"
]

result = embedding.embed_documents(documents)

print(str(result))
```
