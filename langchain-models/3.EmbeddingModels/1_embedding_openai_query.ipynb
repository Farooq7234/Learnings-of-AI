# Create an embedding for one query

An embedding is a list of numbers representing text in a way that can support semantic comparison and search. This lesson embeds one query string with OpenAI.

## Before you begin
Install the project requirements and set `OPENAI_API_KEY` in `.env` or your environment. The embedding request uses the online API and may incur usage charges.

## Steps
1. Import the OpenAI embedding wrapper and load credentials from `.env`.
2. Configure the embedding model. `dimensions=32` requests a compact 32-number vector from the selected model.
3. Call `embed_query()` because this text is a search query rather than a collection of documents.
4. Print the vector. Its numeric values are useful to algorithms, but are not individually human-readable meanings.

## Context
The result represents the meaning of the whole sentence as one vector; it is not a translation or a generated answer. In a search workflow, this query vector is compared with vectors made from stored documents. Those document vectors must use the same embedding model and dimensions so that the comparison is meaningful.

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embedding = OpenAIEmbeddings(model='text-embedding-3-large', dimensions=32)

result = embedding.embed_query("Delhi is the capital of India")

print(str(result))
```
