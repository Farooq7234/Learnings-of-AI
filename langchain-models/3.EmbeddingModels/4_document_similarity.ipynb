# Find the document most similar to a query

This example uses embeddings for a small semantic-search workflow: embed documents and a query, score each document by cosine similarity, then display the highest-scoring match.

## Before you begin
Install the project requirements and set `OPENAI_API_KEY` in `.env` or your environment. The embedding API call may incur usage charges.

## Steps
1. Import the embedding client and cosine-similarity function. NumPy is also imported in the original script, though it is not used.
2. Configure the embedding model and dimensions.
3. Define five cricket-related documents and the natural-language query.
4. Embed the documents together and embed the query with `embed_query()`. Both sides must use the same embedding model and dimensions for a meaningful comparison.
5. `cosine_similarity()` produces one score per document. A larger score means a closer direction in vector space, not a probability.
6. Pair each score with its document index, sort the pairs, select the highest score, and print the query, matching text, and score.

## Context
This is retrieval, not question answering: the code returns the most similar existing sentence and does not ask a language model to compose a response. Cosine similarity compares vector direction, with larger scores indicating a closer match. The selected index connects that score back to the original text in `documents`.

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

load_dotenv()

embedding = OpenAIEmbeddings(model='text-embedding-3-large', dimensions=300)

documents = [
    "Virat Kohli is an Indian cricketer known for his aggressive batting and leadership.",
    "MS Dhoni is a former Indian captain famous for his calm demeanor and finishing skills.",
    "Sachin Tendulkar, also known as the 'God of Cricket', holds many batting records.",
    "Rohit Sharma is known for his elegant batting and record-breaking double centuries.",
    "Jasprit Bumrah is an Indian fast bowler known for his unorthodox action and yorkers."
]

query = 'tell me about bumrah'

doc_embeddings = embedding.embed_documents(documents)
query_embedding = embedding.embed_query(query)

scores = cosine_similarity([query_embedding], doc_embeddings)[0]

index, score = sorted(list(enumerate(scores)),key=lambda x:x[1])[-1]

print(query)
print(documents[index])
print("similarity score is:", score)
```
