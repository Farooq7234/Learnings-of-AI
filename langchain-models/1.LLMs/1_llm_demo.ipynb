# Calling an OpenAI text-completion model

This lesson sends a prompt to an OpenAI model through LangChain and prints the generated text. It uses the text-completion interface (`OpenAI`), rather than the chat-oriented `ChatOpenAI` interface.

## Before you begin

Install the packages listed in `requirements.txt`. Set `OPENAI_API_KEY` in a `.env` file in the working directory (or otherwise in your environment); `load_dotenv()` reads that file. The API call requires internet access and may incur usage charges.

## Steps

1. Import the LangChain OpenAI wrapper and the helper that loads environment variables.
2. Load `.env`, then configure the `gpt-3.5-turbo-instruct` completion model.
3. Send the question with `invoke()`. The returned value is the model's generated text.
4. Print the text so it is visible in the notebook output.

```python
from langchain_openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

llm = OpenAI(model='gpt-3.5-turbo-instruct')

result = llm.invoke("What is the capital of India")

print(result)
```
