# Chat with a Hugging Face hosted model

This lesson uses the Hugging Face Inference API rather than downloading model weights for local inference. LangChain's `ChatHuggingFace` adapts a Hugging Face endpoint to the chat-model interface.

## Before you begin
Install the project requirements and set a Hugging Face access token in `.env` or your environment (commonly `HUGGINGFACEHUB_API_TOKEN`). The model must be available through the selected hosted inference service; access and usage may have provider limits or charges.

## Steps
1. Import the endpoint and chat adapter, then load environment variables.
2. Configure the repository ID and `text-generation` task for the hosted endpoint.
3. Wrap the endpoint with `ChatHuggingFace` so it can accept chat-style inputs.
4. Send the question with `invoke()` and print the returned message's text from `.content`.

```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from dotenv import load_dotenv

load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation"
)

model = ChatHuggingFace(llm=llm)

result = model.invoke("What is the capital of India")

print(result.content)
```
