# Chat with Anthropic Claude

This notebook uses LangChain's Anthropic integration to ask Claude a question and print its response.

## Before you begin
Install the project requirements and set `ANTHROPIC_API_KEY` in `.env` or your environment. The service call requires internet access and may incur API charges.

## Steps
1. Import `ChatAnthropic` and load environment variables from `.env`.
2. Configure the Claude model by its provider model ID. If this model ID is no longer available to your account, choose a currently supported ID from Anthropic's documentation.
3. Use `invoke()` to send the question. The result is a LangChain AI message.
4. Read `.content` to get and print the generated answer as plain text.

```python
from langchain_anthropic import ChatAnthropic
from dotenv import load_dotenv

load_dotenv()

model = ChatAnthropic(model='claude-3-5-sonnet-20241022')

result = model.invoke('What is the capital of India')

print(result.content)
```
