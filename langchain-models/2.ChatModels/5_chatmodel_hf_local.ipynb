# Run a Hugging Face chat model locally

Unlike the hosted endpoint example, this notebook runs the model on this computer. The model weights are downloaded and cached, and inference uses local compute.

## Before you begin
Install the project requirements. The first run needs internet access to download the model; enough disk space and suitable CPU/GPU memory are required. The cache path below is specific to the original author's machine: change `D:/huggingface_cache` to a writable location on your computer if needed.

## Steps
1. Import the chat adapter and local pipeline wrapper, and set the Hugging Face cache directory.
2. Load TinyLlama for the `text-generation` task. `temperature` controls sampling variety; `max_new_tokens` caps newly generated tokens.
3. Wrap the local pipeline with `ChatHuggingFace`.
4. Invoke the model with a question, then print the text in the returned message.

```python
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline
import os

os.environ['HF_HOME'] = 'D:/huggingface_cache'

llm = HuggingFacePipeline.from_model_id(
    model_id='TinyLlama/TinyLlama-1.1B-Chat-v1.0',
    task='text-generation',
    pipeline_kwargs=dict(
        temperature=0.5,
        max_new_tokens=100
    )
)
model = ChatHuggingFace(llm=llm)

result = model.invoke("What is the capital of India")

print(result.content)
```
