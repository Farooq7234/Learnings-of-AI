# Chat with OpenAI

This example sends a request to OpenAI's chat-model interface and displays the assistant's reply.

## Before you begin
Install the project requirements and provide `OPENAI_API_KEY` in `.env` or your environment. Requests need internet access and may incur API charges.

## Steps
1. Import `ChatOpenAI` and load environment variables from `.env`.
2. Create a GPT-4 chat model. `temperature=1.5` asks for more varied output; `max_completion_tokens=10` limits the response to ten generated tokens, so the requested five-line poem may be cut short.
3. Call `invoke()` with the prompt. LangChain returns an AI message; `.content` extracts its text.
4. Print the text.

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv()

model = ChatOpenAI(model='gpt-4', temperature=1.5, max_completion_tokens=10)

result = model.invoke("Write a 5 line poem on cricket")

print(result.content)
```
