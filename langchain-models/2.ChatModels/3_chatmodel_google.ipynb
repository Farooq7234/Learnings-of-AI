# Chat with Google Gemini

This example calls a Gemini chat model through LangChain and prints the model's reply.

## Before you begin
Install the project requirements and provide a Google AI API key, commonly as `GOOGLE_API_KEY` in `.env` or the process environment. The request needs internet access and may incur API charges.

## Steps
1. Import `ChatGoogleGenerativeAI` and load environment variables from `.env`.
2. Create the Gemini model with its provider model ID. If this ID is retired or unavailable to your account, use a currently supported ID.
3. Send the question with `invoke()`, which returns a LangChain AI message.
4. Print `.content`, the text of that message.

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv

load_dotenv()

model = ChatGoogleGenerativeAI(model='gemini-1.5-pro')

result = model.invoke('What is the capital of India')

print(result.content)
```
